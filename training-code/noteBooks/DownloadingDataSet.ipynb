```python
import xarray as xr
import numpy as np
import os

# --------------------------------------------------
# Paths
# --------------------------------------------------

SSH_PATH = "/content/drive/MyDrive/SIH2026/Dataset/SSH_Monthly/SSH_Feb_Mar_2020.nc"

OUTPUT_DIR = "/content/drive/MyDrive/SIH2026/Dataset/SSH_Monthly"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# --------------------------------------------------
# Load SSH
# --------------------------------------------------

ssh_ds = xr.open_dataset(SSH_PATH)

sla = ssh_ds["sla"]

print("Original SLA shape:", sla.shape)

# --------------------------------------------------
# Convert longitude:
# -180 ... 179.875
#        ↓
#   0 ... 359.75
# --------------------------------------------------

sla = sla.assign_coords(
    longitude=(sla.longitude % 360)
)

sla = sla.sortby("longitude")

# --------------------------------------------------
# Target 1° × 1° grid
# --------------------------------------------------

target_lat = xr.DataArray(
    np.arange(-89.5, 90.0, 1.0),
    dims="latitude"
)

target_lon = xr.DataArray(
    np.arange(0.5, 360.0, 1.0),
    dims="longitude"
)

# --------------------------------------------------
# Regrid 0.25° → 1°
# --------------------------------------------------

sla_1deg = sla.interp(
    latitude=target_lat,
    longitude=target_lon,
    method="linear"
)

print("\nRegridded SLA shape:", sla_1deg.shape)

# --------------------------------------------------
# Save February and March separately
# --------------------------------------------------
# --------------------------------------------------
# Save February and March separately
# --------------------------------------------------

# February = first time index
sla_feb = sla_1deg.isel(time=0)

# March = second time index
sla_mar = sla_1deg.isel(time=1)

# Remove time dimension if it is still present
if "time" in sla_feb.dims:
    sla_feb = sla_feb.squeeze("time")

if "time" in sla_mar.dims:
    sla_mar = sla_mar.squeeze("time")


# --------------------------------------------------
# Output paths
# --------------------------------------------------

feb_path = os.path.join(
    OUTPUT_DIR,
    "SSH_202002_1deg.nc"
)

mar_path = os.path.join(
    OUTPUT_DIR,
    "SSH_202003_1deg.nc"
)


# --------------------------------------------------
# Save
# --------------------------------------------------

sla_feb.to_netcdf(feb_path)
sla_mar.to_netcdf(mar_path)

print("Saved successfully:")
print(feb_path)
print(mar_path)
```

```python
import numpy as np

def check_array(name, arr):
    arr = np.asarray(arr)

    print(f"\n===== {name} =====")
    print("Shape :", arr.shape)
    print("NaN   :", np.isnan(arr).sum())
    print("Valid :", np.isfinite(arr).sum())
    print("Min   :", np.nanmin(arr))
    print("Max   :", np.nanmax(arr))
    print("Mean  :", np.nanmean(arr))
    print("Std   :", np.nanstd(arr))


# February
check_array("X February - SST", X_feb[:, :, 0])
check_array("X February - SSH", X_feb[:, :, 1])
check_array("Y February - Argo TEMP", Y_feb)

# March
check_array("X March - SST", X_mar[:, :, 0])
check_array("X March - SSH", X_mar[:, :, 1])
check_array("Y March - Argo TEMP", Y_mar)
```
